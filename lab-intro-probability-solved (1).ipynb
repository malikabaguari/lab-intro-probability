**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 

If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
# Each ticket holder shows up with probability 0.97 -> X ~ Binomial(n=460, p=0.97)
# Everyone has a seat if at most 450 passengers show up: P(X <= 450)
from scipy.stats import binom

n, p_show, seats = 460, 0.97, 450
p_enough_seats = binom.cdf(seats, n, p_show)
print(f"P(a seat for every passenger) = {p_enough_seats:.4f}")
print(f"P(overbooked) = {1 - p_enough_seats:.4f}")
```

**Answer:** The distribution is Binomial(460, 0.97). The probability that every passenger who shows up has a seat is about **88.4%** (so there is roughly an 11.6% chance of overbooking problems).

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
# Number of attempts until the first success -> Geometric(p=0.3)
# At least 3 attempts = the first 2 attempts fail: P(X >= 3) = 1 - P(X <= 2)
from scipy.stats import geom

p = 0.3
p_at_least_3 = geom.sf(2, p)          # sf(k) = P(X > k)
print(f"P(at least 3 attempts) = {p_at_least_3:.4f}")
print("Check by hand:", round((1 - p) ** 2, 4))
```

**Answer:** The number of attempts follows a Geometric distribution. P(at least 3 attempts) = (1 − 0.3)² = **0.49**.

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
# Number of visits per hour -> Poisson(lambda=500)
# Server overwhelmed if more than 550 visits: P(X > 550)
from scipy.stats import poisson

lam, capacity = 500, 550
p_overwhelmed_hour = poisson.sf(capacity, lam)
print(f"P(overwhelmed in one hour) = {p_overwhelmed_hour:.4f}")
```

**Answer:** With a Poisson(500) model, the probability that the server is overwhelmed (more than 550 visits) in a given hour is about **1.29%**.

What is the probability of being overwhelmed at some point during a day? (consider 24hours)

```python
# Over 24 independent hours: P(at least one overwhelmed hour) = 1 - (1 - p)^24
p_overwhelmed_day = 1 - (1 - p_overwhelmed_hour) ** 24
print(f"P(overwhelmed at least once in 24h) = {p_overwhelmed_day:.4f}")
```

**Answer:** Treating the 24 hours as independent, the probability of being overwhelmed at least once during a day is about **26.8%**.

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
# Time between arrivals -> Exponential with mean 10 min (scale = 10, rate = 1/10)
# P(T <= 5)
from scipy.stats import expon

mean_wait = 10
p_within_5 = expon.cdf(5, scale=mean_wait)
print(f"P(next customer arrives within 5 min) = {p_within_5:.4f}")
```

**Answer:** With an Exponential distribution (mean 10 min), the probability that the next customer arrives within 5 minutes is 1 − e^(−0.5) ≈ **39.3%**.

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
# Break if no customer for 15 minutes: P(T > 15)
p_break = expon.sf(15, scale=mean_wait)
print(f"P(employee takes a break) = {p_break:.4f}")
```

**Answer:** The probability of no customer for 15 minutes (so an employee can take a break) is e^(−1.5) ≈ **22.3%**.

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
# Bird weight ~ Normal(mean=150, sd=10): P(140 <= X <= 160)
from scipy.stats import norm

mu, sigma = 150, 10
p_between = norm.cdf(160, mu, sigma) - norm.cdf(140, mu, sigma)
print(f"P(140 <= weight <= 160) = {p_between:.4f}")
```

**Answer:** About **68.3%** (within one standard deviation of the mean, the 68–95–99.7 rule).

If the lifetime (in hours) of a certain electronic component follows an exponential distribution with a mean lifetime of 50 hours, what is the probability that the component fails within the first 30 hours?

```python
# Component lifetime ~ Exponential with mean 50 h (scale = 50): P(T <= 30)
p_fail_30 = expon.cdf(30, scale=50)
print(f"P(fails within the first 30 hours) = {p_fail_30:.4f}")
```
